```python
import csv
import os

# Define the file path and output directory
file_path = 'country_full.csv'  # Make sure the CSV is in your Colab Notebooks folder
output_dir = '.'  # Output directory (where the new region files will be saved)

# Dictionary to hold countries by region
regions = {}

try:
    with open(file_path, newline='') as csvfile: # Open the CSV file
        reader = csv.DictReader(csvfile)

        for row in reader: # Get the country's name and region from the row
            country_name = row['name']
            region = row['region'].strip() if row['region'].strip() else 'Other' # Check if the region is empty

            # Add the country to the appropriate region list
            if region not in regions:
                regions[region] = []
            regions[region].append({'name': country_name, 'region': region})
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
except PermissionError:
    print(f"Error: Permission denied when trying to read '{file_path}'.")
except IOError as e:
    print(f"IO error occurred: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

# Write one file per region
for region, countries in regions.items():
    output_filename = f"{region}.csv"
    output_path = os.path.join(output_dir, output_filename)
    try: # Open a new CSV file for writing
        with open(output_path, mode='w', newline='') as outfile:
            writer = csv.DictWriter(outfile, fieldnames=['name', 'region'])
            writer.writeheader()
            for country in countries:
                writer.writerow(country)
        print(f"Wrote {len(countries)} countries to {output_filename}")
    except PermissionError:
        print(f"Error: Permission denied when trying to write '{output_filename}'.")
    except IOError as e:
        print(f"IO error occurred while writing '{output_filename}': {e}")
```
